```text
            VALOR ENTRADA (Z)
                    │
            ┌───────┴───────┐
            │               │
        ReLU          Sigmoid
            │               │
            ├───────┐       ├───────┐
            │       │       │       │
        Si Z = -2   Si Z = 3  Si Z = -2   Si Z = 3
            │       │       │       │
        0 (Muere)   3 (Pasa)  ~0.12     ~0.95
            │       │       │       │
            ▼       ▼       ▼       ▼
        No frena   Fluye con  Se aplana Se aplana
        el flujo   toda la    (Gradiente (Gradiente
        (Rápido)   potencia   casi cero) casi cero)
            │       │       │       │
            └───────┴───────┴───────┘
                            │
                            ▼
                EL HÉROE vs EL VILLANO
            (ReLU entrena rápido / Sigmoid frena)
```
